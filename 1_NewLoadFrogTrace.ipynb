```python
#Putting my code into a package, FROG.py
import FROG as frog
```

```python
#Loading in files

#Declare an instance of the class:
load_map = frog.loading_mapping()
#Defaults in python package: filename = 'data_1400_afterBS.dat', wave = 'wavelengths_1400_afterBS.dat', time = 'times_1400_afterBS.dat'

data_file = load_map.load_files()[0]
wave_file = load_map.load_files()[1] #Wavelengths (nm)
time_file = load_map.load_files()[2] #Time (fs)
```

```python
#Cleanup function essentially throws everything into a single df with the wavelengths as the row, time as the column. 

new_df = load_map.cleanup(data_file, wave_file, time_file)[0]
new_wave = load_map.cleanup(data_file, wave_file, time_file)[1]
new_time = load_map.cleanup(data_file, wave_file, time_file)[2]

new_df
```

Obtaining the frequency to store it for later use:

```python
freq = load_map.freq(new_wave)
```

Store the variables for later use below:

```python
#Storing the data to be utilized in the next notebooks

%store new_df
%store new_wave
%store new_time
%store freq
```

```python
#heat_map plots out the spectrometer image, ie. Frequency (Hz) vs. Time (fs) with intensity.

load_map.heat_map(new_df, new_wave, new_time,'THz', 'fs', [60, 200],[400, 450], 'plasma', 'Frequency (Hz) vs. Time Delay (fs)')
```
