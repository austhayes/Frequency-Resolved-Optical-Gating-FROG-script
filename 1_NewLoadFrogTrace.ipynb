Within this file (1_NewLoadFrogTrace), the main focus will be to load in the main files obtained from a spectrometer, those files include: intensities, wavelengths, and times. We will then plot out the newly loaded in data in a few steps:

  1. Load in the files utilizing the load_files() function from the FROG.py package.
  2. Cleanup the dataframe using the cleanup() function and setting the time to be the x axis labels and the wavelengths to be the y axis labels within said function. 
  3. Plot the heatmap utilizing the heat_map() function, pay close attention to the inputs.

From here, the result should showcase a heat map like that of which is seen on your spectrometer. 

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import pandas as pd
import math

#Putting my code into a package, FROG.py
import FROG as frog
```

```python
#Loading in files
#Defaults in python package: filename = 'data_1400_afterBS.dat', wave = 'wavelengths_1400_afterBS.dat', time = 'times_1400_afterBS.dat'

data_file = frog.load_files()[0]
wave_file = frog.load_files()[1] #Wavelengths (nm)
time_file = frog.load_files()[2] #Time (fs)
```

```python
#Cleanup function essentially throws everything into a single df with the wavelengths as the row, time as the column. 

new_df = frog.cleanup(data_file, wave_file, time_file)[0]
new_wave = frog.cleanup(data_file, wave_file, time_file)[1]
new_time = frog.cleanup(data_file, wave_file, time_file)[2]

new_df
```

```python
#heat_map plots out the spectrometer image, ie. Frequency (Hz) vs. Time (fs) with intensity.

frog.heat_map(new_df, new_wave, new_time,'Hz', 'fs',[27,222], [4.0*10**5, 4.5*10**5], 'plasma', 'Frequency (Hz) vs. Time (fs)')
```
